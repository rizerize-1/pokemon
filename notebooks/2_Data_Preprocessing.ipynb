## **MERGE DATASET**
Để trực quan hoá một cách hiệu quả bộ dữ liệu về Pokémon, trước hết cần kết hợp thông tin từ cả hai bộ dữ liệu `pokemon.csv` và `pokemonfullstats.csv` để có được bảng dữ liệu đầy đủ về đặc tính của các Pokémon  
Nhưng có một số vấn đề cần phải giải quyết trước khi thực hiện việc này

Cột `Name` có 1 giá trị missing, nếu merge thẳng sẽ gây ra thiếu dữ liệu trận đấu sau này.  
Giả sử tên của các Pokémon sẽ được xếp theo thứ tự ID, hãy xem qua các dòng xung quanh của giá trị bị thiếu này.

```python
missing_point = pokemon.index[pokemon['Name'].isna()][0]
pokemon.loc[missing_point - 3:missing_point + 3, ['Name', 'Type 1']]
```

Như vậy giá trị bị thiếu nằm giữa hai Pokémon có tên 'Mankey' và 'Growlithe', có thể suy luận giá trị từ các web tổng hợp thông tin pokemon, hoặc xem từ `pokemonfullstats.csv` vì dataset này không có giá trị thiếu trên cột `Name`

```python
target_point = pokemonfullstats.index[pokemonfullstats['Name'] == 'Mankey'][0]
pokemonfullstats.loc[target_point - 2:target_point + 4, ['Name', 'Type']]
```

Có thể thấy file `combats` có một số dữ liệu bị lặp, cần loại bỏ giá trị bị lặp đi để không gây ảnh hưởng

Cách đánh số ID của các Pokémon trong hai bộ dữ liệu không đồng nhất, như trong `pokemon.csv`, Charmander có ID ở cột `#` là 5, nhưng trong `pokemonfullstats.csv`, lại có DexNumber là 4, và sự chênh lệch này sẽ lặp lại đến cuối bộ dữ liệu

```python
pd.concat([pokemon[['#', 'Name']].head(10),pokemonfullstats[['DexNumber', 'Name']].head(10)], axis=1)
```

Nguyên nhân gây ra sự chênh lệch này là do trong `pokemonfullstats.csv`, dữ liệu được lấy từ [serebii.net](https://serebii.net/pokemon) và [bulbapedia](https://bulbapedia.bulbagarden.net/), và theo quy ước của các trang web này, các Pokémon tiến hoá ở bậc Mega sẽ nằm cùng một trang với Pokémon gốc của nó và có cùng số Pokedex, rất có thể thay đổi này được thực hiện sau khi bộ dữ liệu `pokemon.csv` được thu thập dẫn đến sự chênh lệch này

Tuy nhiên bộ dữ liệu `pokemon.csv` có cùng nguồn gốc với `combats.csv`, nên để đảm bảo tính nhất quán của dữ liệu, chúng tôi vẫn sẽ sử dụng cách đánh số ID từ `pokemon.csv` làm chuẩn, và merge hai bảng dữ liệu dựa trên cột `Name` 

```python
pokemonfullstats.drop(columns=['DexNumber'], inplace=True)
```

Cả 2 bộ dữ liệu cũng có một số cột có nội dung tương tự nhau, ví dụ như các chỉ số chiến đấu (HP, Attack, Defense, Special Attack, Special Defense, Speed), để tránh việc dư thừa thông tin khi kết hợp hai bảng dữ liệu, chúng tôi chỉ giữ lại các cột từ `pokemonfullstats.csv` loại bỏ các cột tương ứng từ `pokemon.csv`

```python
pokemon.drop(columns=['HP', 'Attack', 'Defense', 'Sp. Atk', 'Sp. Def', 'Speed', 'Generation', 'Legendary'], inplace=True)
```

```python
merged_pokemon = pd.merge(pokemon, pokemonfullstats, left_on='Name', right_on='Name', how='outer')
merged_pokemon = merged_pokemon.dropna(subset=['#'])
merged_pokemon.sort_values(by=['#'], inplace=True, ignore_index=True)
merged_pokemon.rename(columns={'#': 'ID'}, inplace=True)
merged_pokemon.drop(columns=['Type'], inplace = True)
```

### **Missing Value**

```python
missing_values = raw_data.isna().mean()
missing_ratio = missing_values[missing_values > 0].sort_values(ascending=False).to_frame("missing_ratio")
display(missing_ratio.style.background_gradient(cmap="Reds").format("{:.1%}"))
```

```python
key_features = [
    'Type 2', 'Hp', 'Attack', 'Defense', 'SpecialAttack', 'SpecialDefense', 'Speed', 'TotalStats',
    'CatchRate', 'BaseFriendship', 'LevelingRate', 'Weight', 'Height',
    'HasMega', 'TotalEvoStages', 'PreevoName', 'DamageFromPsychic', 'DamageFromIce'
]
selected_missing = missing_ratio[missing_ratio.index.isin(key_features)]
plot_df = selected_missing.reset_index()
plot_df.columns = ['Feature', 'Missing_Ratio']
plot_df = plot_df.sort_values(by='Missing_Ratio', ascending=True)
plt.figure(figsize=(10, 7))
bars = plt.barh(plot_df['Feature'], plot_df['Missing_Ratio'], color='#1f77b4', height=0.8)

plt.title('Top 20 Features with Highest Missing Values', fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Missing Ratio (%)', fontsize=11)
plt.gca().xaxis.set_major_formatter(mtick.PercentFormatter(xmax=1.0))
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.grid(axis='x', linestyle='--', alpha=0.5)

for bar in bars:
    if bar.get_width() > 0: # Chỉ hiện số nếu có missing
        plt.text(bar.get_width() + 0.005, bar.get_y() + bar.get_height()/2, 
                 f'{bar.get_width():.1%}', va='center', fontsize=9)

plt.tight_layout()
plt.show()
```



Vậy là mọi Pokémon bị thiếu thông tin ở trên đều là do tên của chúng không khớp với tên trong `pokemonfullstats.csv`, do đó việc bổ sung thông tin cho các Pokémon này sẽ được thực hiện bằng cách tham khảo nguồn và thêm lại từ `pokemonfullstats.csv`

```python
def replace_data(pokemon_name, pfs_name):
    data = pokemonfullstats[pokemonfullstats['Name'] == pfs_name].iloc[0]
    index = raw_data[raw_data['Name'] == pokemon_name].index[0]
    for col in raw_data.columns:
        if col not in ['ID', 'Name', 'Type 1', 'Type 2']:
            raw_data.loc[index, col] = data[col]

# Unown
replace_data('Unown', 'Unown One Form')
        
# Ho-Oh
replace_data('Ho-oh', 'Ho-Oh')

# Castform, không chú thích nên mặc định sẽ để dạng Normal
replace_data('Castform', 'Castform Normal')

# Burmy có chỉ số gốc giống nhau cho tất cả các dạng
replace_data('Burmy', 'Burmy Plant Cloak')

# Cherrim có chỉ số gốc giống nhau cho tất cả các dạng
replace_data('Cherrim', 'Cherrim Overcast Form')

# Shellos có chỉ số gốc giống nhau cho tất cả các dạng
replace_data('Shellos', 'Shellos West Sea')

# Gastrodon là dạng tiến hoá của Shellos, tương tự
replace_data('Gastrodon', 'Gastrodon West Sea')

# Basculin có chỉ số gốc giống nhau cho tất cả các dạng
replace_data('Basculin', 'Basculin Red-Striped Form')

# Deerling có chỉ số gốc giống nhau cho tất cả các dạng
replace_data('Deerling', 'Deerling Spring Form')

# Sawsbuck là dạng tiến hoá của Deerling, tương tự
replace_data('Sawsbuck', 'Sawsbuck Spring Form')

# Genesect, không chú thích nên mặc định sẽ để dạng No Drive
replace_data('Genesect', 'No Drive Genesect')

# Vivillon
replace_data('Vivillon', 'Vivillon Meadow Pattern')

# Flabébé, các dạng chỉ thay đổi màu sắc, chỉ số gốc giống nhau
replace_data('Flabébé', 'Flabébé Red Flower')

# Floette là dạng tiến hoá của Flabébé, tương tự
replace_data('Floette', 'Floette Red Flower')

# Florges là dạng tiến hoá của Floette, tương tự
replace_data('Florges', 'Florges Red Flower')

# Furfrou
replace_data('Furfrou', 'Furfrou Natural Form')

# Xerneas, dạng Active chỉ thay đổi màu sừng, chỉ số gốc giống nhau
replace_data('Xerneas', 'Xerneas Neutral Mode')

# DeoxysAttack Forme, lỗi chính tả
replace_data('DeoxysAttack Forme', 'Deoxys Attack Forme')

# Kyurem Black Kyurem, Kyurem White Kyurem, cách ghi tên khác nhau
replace_data('Kyurem Black Kyurem', 'Black Kyurem')
replace_data('Kyurem White Kyurem', 'White Kyurem')

# Keldeo Ordinary Forme, Keldeo Resolute Forme, lỗi chính tả
replace_data('Keldeo Ordinary Forme', 'Keldeo Ordinary Form')
replace_data('Keldeo Resolute Forme', 'Keldeo Resolute Form')

# Zygarde Half Forme, cách ghi tên khác nhau
replace_data('Zygarde Half Forme', 'Zygarde 50% Forme')
```

```python
def add_data_mega_primal(pokemon_name, Abilities: str, HiddenAbility: str, Generation: str,
             HP: float, Attack: float, Defense: float, SpecialAttack: float,
             SpecialDefense: float, Speed: float, TotalStats: float,
             Weight: float, Height: float, GenderProbM: str, Category: str,
             CatchRate: float, EggCycles: float, EggGroups: str,
             LevelingRate: str, BaseFriendship: float, IsLegendary: float, IsMythical: float,
             EvoStage: float, TotalEvoStages: float,
             PreevoName: str, DamageFromNormal: float, DamageFromFighting: float,
             DamageFromFlying: float, DamageFromPoison: float, DamageFromGround: float,
             DamageFromRock: float, DamageFromBug: float, DamageFromGhost: float,
             DamageFromSteel: float, DamageFromFire: float, DamageFromWater: float,
             DamageFromGrass: float, DamageFromElectric: float, DamageFromPsychic: float,
             DamageFromIce: float, DamageFromDragon: float, DamageFromDark: float,
             DamageFromFairy: float, HasMega: float = 1.0, IsUltraBeast: float = 0.0
             ):
    index = raw_data[raw_data['Name'] == pokemon_name].index[0]
    raw_data.loc[index, 'Abilities'] = Abilities
    raw_data.loc[index, 'HiddenAbility'] = HiddenAbility
    raw_data.loc[index, 'Generation'] = Generation
    raw_data.loc[index, 'Hp'] = HP
    raw_data.loc[index, 'Attack'] = Attack
    raw_data.loc[index, 'Defense'] = Defense
    raw_data.loc[index, 'SpecialAttack'] = SpecialAttack
    raw_data.loc[index, 'SpecialDefense'] = SpecialDefense
    raw_data.loc[index, 'Speed'] = Speed
    raw_data.loc[index, 'TotalStats'] = TotalStats
    raw_data.loc[index, 'Weight'] = Weight
    raw_data.loc[index, 'Height'] = Height
    raw_data.loc[index, 'GenderProbM'] = GenderProbM
    raw_data.loc[index, 'Category'] = Category
    raw_data.loc[index, 'CatchRate'] = CatchRate
    raw_data.loc[index, 'EggCycles'] = EggCycles
    raw_data.loc[index, 'EggGroup'] = EggGroups
    raw_data.loc[index, 'LevelingRate'] = LevelingRate
    raw_data.loc[index, 'BaseFriendship'] = BaseFriendship
    raw_data.loc[index, 'IsLegendary'] = IsLegendary
    raw_data.loc[index, 'IsMythical'] = IsMythical
    raw_data.loc[index, 'IsUltraBeast'] = IsUltraBeast
    raw_data.loc[index, 'HasMega'] = HasMega
    raw_data.loc[index, 'EvoStage'] = EvoStage
    raw_data.loc[index, 'TotalEvoStages'] = TotalEvoStages
    raw_data.loc[index, 'PreevoName'] = PreevoName
    raw_data.loc[index, 'DamageFromNormal'] = DamageFromNormal
    raw_data.loc[index, 'DamageFromFighting'] = DamageFromFighting
    raw_data.loc[index, 'DamageFromFlying'] = DamageFromFlying
    raw_data.loc[index, 'DamageFromPoison'] = DamageFromPoison
    raw_data.loc[index, 'DamageFromGround'] = DamageFromGround
    raw_data.loc[index, 'DamageFromRock'] = DamageFromRock
    raw_data.loc[index, 'DamageFromBug'] = DamageFromBug
    raw_data.loc[index, 'DamageFromGhost'] = DamageFromGhost
    raw_data.loc[index, 'DamageFromSteel'] = DamageFromSteel
    raw_data.loc[index, 'DamageFromFire'] = DamageFromFire
    raw_data.loc[index, 'DamageFromWater'] = DamageFromWater
    raw_data.loc[index, 'DamageFromGrass'] = DamageFromGrass
    raw_data.loc[index, 'DamageFromElectric'] = DamageFromElectric
    raw_data.loc[index, 'DamageFromPsychic'] = DamageFromPsychic
    raw_data.loc[index, 'DamageFromIce'] = DamageFromIce
    raw_data.loc[index, 'DamageFromDragon'] = DamageFromDragon
    raw_data.loc[index, 'DamageFromDark'] = DamageFromDark
    raw_data.loc[index, 'DamageFromFairy'] = DamageFromFairy
```


Kiểu dữ liệu của ID được chuyển về string để bỏ dấu thập phân ở cuối


```python
raw_data['ID'] = raw_data['ID'].astype(str)
raw_data['ID'] = raw_data['ID'].str.replace(r'\.0$', '', regex=True)
```


Kiểu dữ liệu của một số cột nên được chuyển sang int để tránh rắc rối sau này


```python
for col in ['CatchRate', 'EggCycles', 'BaseFriendship', 'IsLegendary', 'IsMythical', 'IsUltraBeast', 'HasMega', 'EvoStage', 'TotalEvoStages']:
    raw_data[col] = raw_data[col].astype(str)
    raw_data[col] = raw_data[col].str.replace(r'\.0$', '', regex=True)
    raw_data[col] = raw_data[col].astype(int)
```


Các dữ liệu trống của cột `GenderProbM` và `EggGroup` cũng nên được điền lại để thuận tiện cho việc xử lý sau này


```python
raw_data.loc[raw_data['GenderProbM'] == '-', 'GenderProbM'] = -1 # Pokémon vô tính
raw_data.loc[raw_data['EggGroup'] == '-', 'EggGroup'] = '[]'
```

Cột `LevelingRate` có một ký tự xuống dòng thừa ở cuối mỗi giá trị, cần loại bỏ ký tự này để tránh lỗi khi xử lý dữ liệu sau này

```python
raw_data['LevelingRate'] = raw_data['LevelingRate'].str.rstrip('\r\n')
```
